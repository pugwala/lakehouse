## What you can query

Each row is a view (a data product) or a cube. Measures are asked for with `MEASURE(name)`; everything else is a dimension you can group or filter by.

```python
%sql SELECT table_name FROM information_schema.tables WHERE table_schema = 'public' ORDER BY 1
```

## The busiest villains, by common name

`people_involved` counts distinct people, so copycats who share a name still count separately. `participations` counts each person once per fight.

```python
%%sql
SELECT common_name, MEASURE(participations) AS fights, MEASURE(people_involved) AS people
FROM fight_participants
WHERE person_role = 'villain' AND common_name IS NOT NULL
GROUP BY 1
ORDER BY 2 DESC
LIMIT 10
```

## Damage by kind of fight

`incident_id` traces damage to one fight; `incident_type` is its kind.

```python
%%sql
SELECT incident_type, MEASURE(damage_cost_usd) AS cost_usd, MEASURE(uninsured_cost_usd) AS uninsured_usd
FROM incident_damage
GROUP BY 1
ORDER BY 2 DESC
```
